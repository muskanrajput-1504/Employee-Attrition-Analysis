### Duplicate Record Check

Duplicate records can lead to inaccurate analysis by counting the same employee more than once. Therefore, the dataset was checked for duplicate rows before proceeding.

```python
missing_values = df.isnull().sum()

missing_values
```

```python
missing_percentage = (df.isnull().sum()/len(df))*100

pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": missing_percentage
})
```
